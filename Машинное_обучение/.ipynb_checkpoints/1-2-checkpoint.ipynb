```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Исходные данные
X = np.array([3,5,7,6,9])
Y = np.array([3, 7, 11, 14, 15])
display(X, Y)
```

```python
plt.figure()
plt.scatter(X, Y)
plt.xlabel("Расходы на рекламу")
plt.ylabel("Объемы продаж")
plt.show()
```

```python
# Считаем оценку эффективную парную деятельность по формуле
a1 = ((X - X.mean())*(Y - Y.mean())).mean()/((X - X.mean())**2).mean()
a0 = Y.mean() - a1 * X.mean()
print(f"Модель линейной регрессии: Y^ = {a0} + {a1} * x")
```

```python
x_space = np.linspace(3, 9, 7)
print(x_space)
y_pred = a0 + a1 * x_space
print(y_pred)
```

```python
fig = plt.figure()
ax = fig.add_axes([0.1, 0.1, 0.8, 0.8])
ax.scatter(X, Y)
ax.plot(x_space, y_pred, 'r')
```
